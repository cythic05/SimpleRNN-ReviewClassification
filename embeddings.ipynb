```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = [ 'the glass of milk',
        'the glass of juice',
        'the cup of tea',
        'I am a good boy',
        'I am a good developer',
        'understand the meaning of words',
        'your videos are good',

]
```

```python
sent
```

```python
vocab_size = 10000
```

```python
ohe = [one_hot(words, vocab_size) for words in sent]
ohe
```

#### Above is the representation for one_hot_encoding <br>
#### Below is the representation for WordEmbedding

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_len = 8
```

```python
embedded_sent = pad_sequences(ohe, maxlen=sent_len, padding='pre')
```

```python
embedded_sent
```

```python
## feature representation

dim = 10
```

```python
model = Sequential()
model.add(Embedding(vocab_size, dim, input_length=sent_len))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
X = model.predict(embedded_sent)
```
